Partie 1 : Chargement et structuration des données (45 min) :

L'objectif de cette étape est de définir les données d'entrée de l'entreprise sous forme de structures Python appropriées.
 Stock initial : Créez un dictionnaire stock dont chaque clé est un identifiant de produit (ex: "P01") et la valeur est un
dictionnaire contenant :
1. nom (ex: "Ecran 24 pouces")
2. prix_unitaire_ht (float)
3. quantite_dispo (int)
4. seuil_alerte (int)
Entrez au moins 5 produits différents.
 Registre des ventes du jour : Créez une liste de tuples ventes_jour représentant les transactions sous la forme (id_produit,
quantite_vendue).
