Partie 2 : Développement du moteur de règles et fonctions (1h00):Pensez à découper votre code en fonctions réutilisables.

Développez les fonctions logiques indispensables aux calculs métier 

Fonction calculer_montant_ttc(prix_ht, quantite):

    1.Calcule le montant HT total.
    2.Applique la TVA standard de 20%.
    3.Applique une remise commerciale de 10%si le montant HT total dépasse 500 €.
    4.Retourne le montant TTC final arrondi à 2 décimales.

Fonction traiter_transaction(id_produit, quantite):

    1.Vérifie si l'identifiant existe dans le stock.
    2.Vérifie si la quantité demandée est disponible en stock.
    3.Si le stock est suffisant : décrémente la quantité du stock et renvoie le montant TTC calculé.
    4.Sinon : affiche un message d'erreur clair et annule la transaction.

```python
def calculer_montant_ht(prix_unitaire_ht, quantite):
    return prix_unitaire_ht * quantite

def remise_10_pourcent(montant_ttc):
    if montant_ttc > 500:
        return round(montant_ttc * 0.9, 2)
    return montant_ttc

def calculer_montant_ttc(prix_ht, quantite):
    montant_ht = calculer_montant_ht(prix_ht, quantite)
    tva = 0.2  # TVA de 20%
    montant_ttc = montant_ht * (1 + tva)
    montant_ttc = remise_10_pourcent(montant_ttc)
    return round(montant_ttc, 2)

def traiter_transaction(code_produit, quantite_vendue):
    if code_produit in stock:
        produit = stock[code_produit]
        if quantite_vendue <= produit["quantite_dispo"]:
            montant_ttc = calculer_montant_ttc(produit["prix_unitaire_ht"], quantite_vendue)
            produit["quantite_dispo"] -= quantite_vendue
            print(f"Vente de {quantite_vendue} unités de {produit['nom']}. Montant TTC: {montant_ttc}€")
            if produit["quantite_dispo"] < produit["seuil_alerte"]:
                print(f"Alerte: Stock faible pour {produit['nom']} (Quantité restante: {produit['quantite_dispo']})")
        else:
            print(f"Erreur: Quantité demandée ({quantite_vendue}) supérieure à la quantité disponible ({produit['quantite_dispo']}) pour {produit['nom']}.")
    else:
        print(f"Erreur: Produit avec le code {code_produit} non trouvé dans le stock.")
```
